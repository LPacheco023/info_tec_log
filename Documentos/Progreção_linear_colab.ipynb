```python
# Importando bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Criando o DataFrame
data = {
    'Experience Years': [1.1, 2.2, 3.1, 3.2, 3.2, 3.7, 3.9, 4.0, 4.1, 4.5,
                         4.9, 5.1, 5.3, 5.5, 5.9, 6.0, 6.2, 6.5, 6.8, 7.1,
                         7.9, 8.2, 8.7, 9.1, 9.5, 9.6, 10.3, 10.5],
    'Salary': [39343, 42774, 46205, 37731, 43525, 39891, 56642, 60150, 54445, 61111,
               67938, 66029, 83088, 82200, 81363, 93940, 91000, 91738, 98273, 101302,
               113812, 111612, 109431, 105582, 116969, 112635, 122391, 121872]
}

df = pd.DataFrame(data)

# Separando as variáveis
X = df[['Experience Years']]
y = df['Salary']

# Criando o modelo
model = LinearRegression()
model.fit(X, y)

# Coeficientes
print(f"Coeficiente angular (b1): {model.coef_[0]:.2f}")
print(f"Intercepto (b0): {model.intercept_:.2f}")

# Fazendo previsões
y_pred = model.predict(X)

# Visualizando o gráfico
plt.scatter(X, y, color='blue', label='Dados reais')
plt.plot(X, y_pred, color='red', label='Regressão Linear')
plt.title('Relação entre Experiência e Salário')
plt.xlabel('Anos de Experiência')
plt.ylabel('Salário')
plt.legend()
plt.show()
```
